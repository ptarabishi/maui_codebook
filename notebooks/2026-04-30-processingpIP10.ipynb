```python
import glob
import os
from src import loader, moco
import numpy as np
import nibabel as nib
```

```python
exp_id = 'fly09_win01*'
base_dir = '/Volumes/AhmedLab/princess/data/'

raw_path = io.get_base_path(exp_id,'raw')
base_name = os.path.basename(raw_path)
print(base_name)

processed_path = io.get_base_path(exp_id,'processed')
print(processed_path)
```

```python
fly_list = ['11']
windows = np.arange(1,4)
print(windows)

exps_to_moco = []
for fly in fly_list:
    for window in windows:
        exp_id = f'fly{fly}_win0{window}*'
        raw_path = io.get_base_path(exp_id,'raw')
        base_name = os.path.basename(raw_path)
        exps_to_moco.append(raw_path)
        processed_directory = os.path.join(base_dir,'processed', base_name)
        if not os.path.exists(processed_directory):
            os.makedirs(processed_directory)

        processed_path = io.get_base_path(exp_id,'processed')

print(exps_to_moco)
```

```python
volumes = 100
for exp in exps_to_moco:
    print(exp)
    func_channel = glob.glob(os.path.join(exp, '*channel_2.nii'))[0]
    print(func_channel)

    img = nib.load(func_channel)
    data_obj = img.dataobj
    func_data = np.asarray(data_obj)
    # func_data = io.load_nii(func_channel)
    print(f'generating fixed_brain from {volumes} volumes')
    mean_brain, fixed_brain = moco.generate_fixed(func_data, volumes)
    moco_func_brain = moco.motion_correction(func_data, fixed_brain)

    moco_nii_name = f'{processed_path}/motion_corrected.nii'
    io.save_nii(moco_nii_name, moco_func_brain)

    if os.path.exists(moco_nii_name):
        print(f'motion corrected data saved to {moco_nii_name}')
```
