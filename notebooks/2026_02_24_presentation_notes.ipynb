```python
# pull timestamp data
def pull_timestamps_ms(time_array, signal_array, threshold):
    binarized_signal = []
    for signal in time_array:
        if signal > threshold:
            binarized_signal.append(1)
        else:
            binarized_signal.append(0)

    # get frame starts by finding +1 changes
    # add [0] because diff loses the first value
    signal_changes = [0] + np.diff(binarized_signal)

    timestamps = []
    for i, val in enumerate(signal_changes):
        if val == 1:
            timestamps.append(signal_array[i])

    return timestamps, signal_changes
```

```python
# pull camera timestamps
cam_ttl = voltage_data[' Input 2']
time = voltage_data['Time(ms)']

# binarize signal
threshold = 3.2 # Volts
binarized_cam = []
for signal in cam_ttl:
    if signal > threshold:
        binarized_cam.append(1)
    else:
        binarized_cam.append(0)

# get frame starts by finding +1 changes
diffs = [0] + np.diff(binarized_cam)

timestamps = []
for i, val in enumerate(diffs):
    if val == 1:
        timestamps.append(time[i])

print(len(timestamps))
```
